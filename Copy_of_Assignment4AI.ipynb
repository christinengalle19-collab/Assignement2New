```python
# Task 9: Iterators and Generators
# Create an iterator for the first 5 even numbers.
# Then, create a generator for the same.

# Your code here:

# Iterator for the first 'limit' even numbers
class EvenNumberIterator:
    def __init__(self, limit):
        self.num = 0
        self.count = 0
        self.limit = limit

    def __iter__(self):
        return self

    def __next__(self):
        if self.count < self.limit:
            even = self.num
            self.num += 2
            self.count += 1
            return even
        else:
            raise StopIteration

# Apply iterator for the first 5 even numbers
print("Iterator output:")
for n in EvenNumberIterator(limit=5):
    print(n)


# Generator for the first 'limit' even numbers
def even_number_generator(limit):
    num = 0
    count = 0
    while count < limit:
        yield num
        num += 2
        count += 1

# Apply generator for the first 5 even numbers
print("\nGenerator output:")
for n in even_number_generator(limit=5):
    print(n)
```

```python
# Task 10: Map, Reduce, and Filter
# Use map to square all numbers in a list.
# Use reduce to find the product of all numbers in a list.
# Use filter to get only even numbers from a list.

numbers = [1, 2, 3, 4, 5]

# Your code here:
# Use map to square all numbers in a list.
squared_numbers = list(map(lambda x: x**2, numbers))

# Use reduce to find the product of all numbers in a list.
from functools import reduce
product = reduce(lambda x, y: x*y, numbers)

# Use filter to get only even numbers from a list.
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
```

```python
try:
    print(squared_numbers)
    print(product)
    print(even_numbers)
except NameError:
    print("squared_numbers, product, or even_numbers is not defined")
```

```python
# Task 11: Object-Oriented Programming - Creating a Class
# Create a class called 'Rectangle' with attributes 'length' and 'width'.
# Include methods to calculate area and perimeter.

# Your code here:
# Create a class called 'Rectangle' with attributes 'length' and 'width'.
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)
```

```python
try:
    print(Rectangle(5, 10).area())
    print(Rectangle(5, 10).perimeter())
except NameError:
    print("Rectangle is not defined")
```
